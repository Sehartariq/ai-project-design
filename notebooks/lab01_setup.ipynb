# AI Project Design & Development — Lab 01

## Project Title
**Real-Time Object Detection System for Smart Surveillance**

---

## Problem Statement
Traditional surveillance systems rely on human monitoring, which is
error-prone and inefficient at scale. There is a need for an automated,
AI-driven system capable of detecting and classifying objects in
real-time video streams with high accuracy and low latency.

---

## Core Objectives
- **Build** a real-time object detection pipeline using YOLO (ultralytics).
- **Process** live video frames with OpenCV at ≥ 15 FPS.
- **Evaluate** model performance using mAP, precision, and recall.
- **Deploy** a lightweight inference module suitable for edge devices.

---
